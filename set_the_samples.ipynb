```python
import pandas as pd
```

```python
df = pd.read_csv('AREA_IMOVEL_1_MICRORREGIOESCSV.csv', encoding='cp1252')
```

```python
df.head()
```

```python
df['the_geom']
```

```python
df_sample = df[['cod_imovel', 'municipio', 'microrregiao']]
```

```python
microrregioes_lista = [
    'Central Sul', 
    'Noroeste', 
    'Centro-Oeste',
    'Nordeste',
    'Litoral Sul', 
    'Rio Doce', 
    'Central Serrana', 
    'Caparao',
    'Metropolitana', 
    'Sudoeste Serrana']
```

```python
df_sample.head()
```

```python
df_sample['cod_imovel'] = df_sample['cod_imovel'].map(lambda x: f"'{x}'")
```

```python
df_sample.head()
```

```python
df_aux = []

for locais in microrregioes_lista:
    # df_sample['cod_imovel'] = df_sample['cod_imovel'].map(lambda x: f"'{x}'")
    sample_1 = df_sample.loc[df_sample['microrregiao']==locais]['cod_imovel'].sample(n=1000, random_state=42)

    # sample_1['cod_imovel'] = sample_1['cod_imovel'].apply(lambda x: f"'{x}'")
    # linha_txt = ', '.join(sample_1)
    sample_1.to_csv(f'imoveis/{locais}.csv', index=False, sep=';')

    # sample_1 = sample_1.assign(microrregiao=locais)

    df_aux.append(sample_1)

df_final = pd.concat(df_aux, ignore_index=True)

df_final.to_csv('imoveis/amostra_total.csv', index=False, sep=';')


    # with open(f"{locais}.txt", 'w', encoding='utf-8') as f:
    #     f.write(linha_txt)
```

```python
df_car = pd.read_csv("planilha_car.csv", sep=";")
```

```python
df_car.head()
```

```python
samples_car = df_car.merge(df_final, how="inner", on="cod_imovel")
samples_car.to_csv('imoveis/amostras_car.csv', sep=";", index=False)
```
