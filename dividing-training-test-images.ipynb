```python
import os
import random

folder_path = "/data/integracar/amostras_car_orotofoto"

image_files = [f for f in os.listdir(folder_path) if f.lower().endswith(('.jpg','.jpeg','.png','.bmp','.tif','.tiff'))]
random.shuffle(image_files)

train_files = image_files[:7000]
test_files = image_files[7000:8500]
validation_files = image_files[8500:]

with open("train.txt", "w") as train_f:
    for img in train_files:
        train_f.write(os.path.join(folder_path, img) + "\n")

with open("validation.txt", "w") as validation_f:
    for img in validation_files:
        validation_f.write(os.path.join(folder_path, img) + "\n")

with open("test.txt", "w") as test_f:
    for img in test_files:
        test_f.write(os.path.join(folder_path, img) + "\n")
```
