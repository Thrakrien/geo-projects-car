```python
from tqdm import tqdm
import glob
import rasterio
import numpy as np

legend_rgb = {
    (150, 150, 150): 1, #Afloramento Rochoso
    (251, 154, 153): 2, #Área Edificada
    (69, 175, 213): 3, #Brejo
    (150, 109, 207): 4,#Campo Rupestre/Altitude
    (128, 214, 16): 5, #Cultivo Agrícola - Abacaxi
    (247, 223, 8): 6, #Cultivo Agrícola - Banana
    (119, 9, 29): 7, #Cultivo Agrícola - Café
    (209, 163, 117): 8, #Cultivo Agrícola - Cana-De-Açúcar
    (231, 67, 97): 9, #Cultivo Agrícola - Coco-Da-Baía
    (245, 141, 23): 10, #Cultivo Agrícola - Mamão
    (55, 196, 201): 11, #Cultivo Agrícola - Outros Cultivos Permanentes
    (225, 175, 38): 12, #Cultivo Agrícola - Outros Cultivos Temporários
    (81, 77, 77): 13, #Extração Mineração
    (211, 127, 122): 14, #Macega
    (156, 68, 203): 15, #Mangue
    (133, 196, 221): 16, #Massa D'Água
    (13, 103, 19): 17, #Mata Nativa
    (51, 160, 44): 18, #Mata Nativa em Estágio Inicial de Regeneração
    (31, 205, 170): 19, #Outros
    (178, 214, 32): 20, #Pastagem
    (207, 103, 65): 21, #Reflorestamento - Eucalipto
    (243, 184, 129): 22, #Reflorestamento - Pinus
    (151, 132, 233): 23, #Reflorestamento - Seringueira
    (63, 231, 161): 24, #Restinga
    (245, 222, 193): 25 #Solo Exposto
}

# Dicionário reverso para salvar legenda legível
legend_names = {
    1: "Afloramento Rochoso",
    2: "Área Edificada",
    3: "Brejo",
    4: "Campo Rupestre/Altitude",
    5: "Cultivo Agrícola - Abacaxi",
    6: "Cultivo Agrícola - Banana",
    7: "Cultivo Agrícola - Café",
    8: "Cultivo Agrícola - Cana-De-Açúcar",
    9: "Cultivo Agrícola - Coco-Da-Baía",
    10: "Cultivo Agrícola - Mamão",
    11: "Cultivo Agrícola - Outros Cultivos Permanentes",
    12: "Cultivo Agrícola - Outros Cultivos Temporários",
    13: "Extração Mineração",
    14: "Macega",
    15: "Mangue",
    16: "Massa D'Água",
    17: "Mata Nativa",
    18: "Mata Nativa em Estágio Inicial de Regeneração",
    19: "Outros",
    20: "Pastagem",
    21: "Reflorestamento - Eucalipto",
    22: "Reflorestamento - Pinus",
    23: "Reflorestamento - Seringueira",
    24: "Restinga",
    25: "Solo Exposto"
}

# Inicializa contagem total por classe
class_counts = {class_id: 0 for class_id in legend_names}
total_pixels = 0

mask_list = glob.glob('/data/integracar/amostras_car/*.tif')  # amostras_car_mask

# mask_list = glob.glob("/data/integracar/amostras_car/ES-3205309-F5D0B0D86B3E413F9EB98146A1725193.tif")

for mask_path in tqdm(mask_list, desc="Processando imagens"):
    with rasterio.open(mask_path) as src:
        img = src.read()  # (C, H, W) para tif
    img_rgb = np.transpose(img[:3], (1, 2, 0))  # Para RGB

    # Conta pixels por RGB
    for rgb, class_id in legend_rgb.items():
        mask = np.all(img_rgb == rgb, axis=-1)
        count = np.sum(mask)
        class_counts[class_id] += count
        total_pixels += count


# for class_id, count in class_counts.items():
#     perc = 100 * count / total_pixels if total_pixels > 0 else 0
#     print(f"{legend_names[class_id]} (Classe {class_id}): {perc:.2f}%")

# print(f"Total de pixels: {total_pixels}")

output_path = "contagem_classes.txt"

with open(output_path, "w", encoding="utf-8") as f:
    for class_id, count in class_counts.items():
        perc = 100 * count / total_pixels if total_pixels > 0 else 0
        line = f"{legend_names[class_id]} (Classe {class_id}): {perc:.2f}% - {count} pixels\n"
        print(line, end="")
        f.write(line)

    total_line = f"Total de pixels: {total_pixels}\n"
    print(total_line, end="")
    f.write(total_line)
```
